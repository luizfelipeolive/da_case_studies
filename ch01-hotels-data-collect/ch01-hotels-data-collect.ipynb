```python
# Selecionar os preços válidos
precos = df["price"].dropna()

# Média de todos os hotéis
media_completa = precos.mean()

# Tamanhos solicitados pelo exercício
tamanhos = [25, 50, 200]

# Lista que receberá os resultados
resultados = []

# Repetir o exercício três vezes
for repeticao in range(1, 4):

    for tamanho in tamanhos:

        amostra = precos.sample(
            n=tamanho,
            replace=False,
            random_state=repeticao
        )

        media_amostra = amostra.mean()

        resultados.append({
            "repeticao": repeticao,
            "tamanho_amostra": tamanho,
            "media_amostra": media_amostra,
            "media_completa": media_completa,
            "diferenca": media_amostra - media_completa
        })

# Transformar os resultados em uma tabela
resultados_df = pd.DataFrame(resultados)

# Arredondar para duas casas decimais
resultados_df = resultados_df.round(2)

resultados_df
```
