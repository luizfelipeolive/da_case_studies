## Dados da cafeteria

A cafeteria registrou as seguintes vendas:

| Produto | Quantidade | Preço unitário |
|:--|--:|--:|
| Café | 120 | R\$ 6,00 |
| Cappuccino | 80 | R\$ 9,00 |
| Chá | 50 | R\$ 5,00 |
| Bolo | 40 | R\$ 12,00 |

### Pergunta

Qual produto gerou o maior faturamento?

O faturamento de cada produto será calculado por:

$$
\text{Faturamento}
=
\text{Quantidade}
\times
\text{Preço unitário}
$$

```python
produtos = ["Café", "Cappuccino", "Chá", "Bolo"]
quantidades = [120, 80, 50, 40]
precos = [6, 9, 5, 12]

faturamentos = [
    quantidade * preco
    for quantidade, preco in zip(quantidades, precos)
]

faturamentos
```

## Interpretação dos resultados

Os faturamentos foram:

| Produto | Faturamento |
|:--|--:|
| Café | R\$ 720,00 |
| Cappuccino | R\$ 720,00 |
| Chá | R\$ 250,00 |
| Bolo | R\$ 480,00 |

Café e cappuccino apresentaram o mesmo faturamento.

### Primeira conclusão

O produto com maior preço não foi necessariamente aquele com maior faturamento.

### Segunda conclusão

O faturamento depende simultaneamente de:

- quantidade vendida;
- preço unitário.

## Representação matemática

Para cada produto $i$, podemos definir:

$$
F_i = q_i p_i
$$

em que:

- $F_i$ é o faturamento;
- $q_i$ é a quantidade vendida;
- $p_i$ é o preço unitário.

O faturamento total é:

$$
F_{\text{total}}
=
\sum_{i=1}^{n} q_i p_i
$$

```python
faturamento_total = sum(faturamentos)

print(f"Faturamento total: R$ {faturamento_total:,.2f}")
```

## Organizando os dados com pandas

Agora construiremos uma tabela usando um `DataFrame`.

```python
import pandas as pd

dados = pd.DataFrame({
    "Produto": produtos,
    "Quantidade": quantidades,
    "Preço": precos,
    "Faturamento": faturamentos
})

dados
```

```python
dados["Participação (%)"] = (
    dados["Faturamento"]
    / dados["Faturamento"].sum()
    * 100
)

dados
```

## Visualização dos faturamentos

O gráfico permite comparar rapidamente os produtos.

```python
import matplotlib.pyplot as plt

dados.plot(
    x="Produto",
    y="Faturamento",
    kind="bar",
    legend=False
)

plt.title("Faturamento por produto")
plt.ylabel("Faturamento em reais")
plt.xlabel("Produto")
plt.xticks(rotation=0)
plt.show()
```

Neste momento, comentar com os alunos que café e cappuccino tiveram o mesmo
faturamento, embora apresentem preços e quantidades diferentes.

## Exercício rápido

Considere um novo produto:

| Produto | Quantidade | Preço |
|:--|--:|--:|
| Sanduíche | 35 | R\$ 15,00 |

Calcule:

1. o faturamento do sanduíche;
2. o novo faturamento total;
3. sua participação percentual no faturamento.

```python
quantidade_sanduiche = 35
preco_sanduiche = 15

faturamento_sanduiche = quantidade_sanduiche * preco_sanduiche
novo_total = faturamento_total + faturamento_sanduiche

participacao = faturamento_sanduiche / novo_total * 100

print("Faturamento do sanduíche:", faturamento_sanduiche)
print("Novo faturamento total:", novo_total)
print(f"Participação: {participacao:.2f}%")
```
